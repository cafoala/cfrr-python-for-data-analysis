# Day 2 checkpoint: yesterday's cleaned Oxford data

This notebook repeats Day 1's cleaning steps in one place, so everyone starts Day 2 with the same `oxford` DataFrame. The Day 2 notebook runs it for you with `%run`. You never need to edit it, but it is worth reading once: it is the whole of yesterday in about a dozen lines.

*Data: Met Office historic station data for Oxford. Contains public sector information licensed under the [Open Government Licence v3.0](https://www.nationalarchives.gov.uk/doc/open-government-licence/version/3/).*

```python
from pathlib import Path
from urllib.request import urlretrieve

import numpy as np
import pandas as pd

BASE = "https://raw.githubusercontent.com/cafoala/cfrr-python-for-data-analysis/snapshot-2026-10-29/"
DATA = Path("data")
DATA.mkdir(exist_ok=True)
# Snapshot of https://www.metoffice.gov.uk/pub/data/weather/uk/climate/stationdata/oxforddata.txt
OXFORD = DATA / "oxforddata.txt"
urlretrieve(BASE + "data/raw/oxforddata.txt", OXFORD)
COLUMNS = ["yyyy", "mm", "tmax", "tmin", "af", "rain", "sun", "status"]
MEASURES = ["tmax", "tmin", "af", "rain", "sun"]
```

```python
oxford = pd.read_csv(OXFORD, sep=r"\s+", skiprows=7, names=COLUMNS, na_values="---")
for col in MEASURES:
    text = oxford[col].astype(str)
    oxford[col + "_estimated"] = text.str.endswith("*")
    oxford[col] = text.str.rstrip("*").astype(float)
oxford["provisional"] = oxford["status"] == "Provisional"
oxford = oxford.drop(columns="status")
oxford["tmean"] = (oxford["tmax"] + oxford["tmin"]) / 2
oxford.index = pd.to_datetime({"year": oxford["yyyy"], "month": oxford["mm"], "day": 1})
```
